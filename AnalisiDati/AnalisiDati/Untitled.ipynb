```python
#importo le libreirie necessarie per analizzare i dati
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
#importo i dati dal csv e creo il mio data_set locale
players_df= pd.read_csv('players.csv')
players_df=players_df[['player_id','name','last_season','country_of_citizenship','date_of_birth', 'sub_position', 'position',
       'foot', 'height_in_cm', 'market_value_in_eur', 'highest_market_value_in_eur','agent_name', 'current_club_name']]
players_df=players_df.drop_duplicates()
players_df=players_df.dropna()
players_df
```

```python
players_df.to_csv('Players_Nona.csv', index=False)
```

```python
competitions_df= pd.read_csv('competitions.csv')
competitions_df=competitions_df[['competition_id', 'name', 'type', 'country_name', 'domestic_league_code']]
competitions_df=competitions_df.drop_duplicates()
competitions_df=competitions_df.dropna()
competitions_df
```

```python
competitions_df.to_csv('competitions_Nona.csv', index=False)
```

```python
club_df= pd.read_csv('clubs.csv')
club_df=club_df[['club_id','name','domestic_competition_id','squad_size','average_age','foreigners_number','foreigners_percentage', 'national_team_players','stadium_name','net_transfer_record']]
club_df=club_df.drop_duplicates()
club_df=club_df.dropna()
club_df = club_df[club_df['net_transfer_record'] != '+-0']
club_df
```

```python
club_df.to_csv('clubs_Nona.csv', index=False)
```

```python
#importo i dati dal csv e creo il mio data_set locale
appearances_df= pd.read_csv('appearances.csv')
appearances_df=appearances_df.drop_duplicates()
appearances_df.dropna(subset=['appearance_id'])
appearances_df
```

```python
appearances_df.to_csv('appearances_Nona.csv', index=False)
```

```python
# Trova le righe dove appearances_id è null
null_appearances = appearances_df[appearances_df['appearance_id'].isnull()]

# Visualizza le righe con appearances_id null
null_appearances
```

```python
null_appearances = appearances_df[appearances_df['appearance_id'].isnull()]

# Righe con appearances_id duplicato (escludendo i null)
duplicate_appearances = appearances_df[appearances_df['appearance_id'].notnull() & appearances_df.duplicated('appearance_id', keep=False)]

# Combina entrambe le condizioni
problematic_appearances = pd.concat([null_appearances, duplicate_appearances])

# Visualizza le righe problematiche
problematic_appearances
```

```python
# Trova i duplicati in appearances_id (escludendo i null)
duplicate_appearances = appearances_df[appearances_df['appearance_id'].notnull() & appearances_df.duplicated('appearance_id', keep=False)]

# Visualizza le righe con appearances_id duplicato
duplicate_appearances
```
